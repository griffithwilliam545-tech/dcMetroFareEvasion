```python
import geopandas as gpd
from pathlib import Path
import pandas as pd
pd.set_option('display.max_rows', None) # so I can see all rows of a df
pd.set_option('display.max_columns', None)
```

```python
df = gpd.read_file('Data/Output_data/metroLines_merge_SpatialJoin_ExportFeatures.shp')
df.head(5)
```

```python
agg_rules = {
    col: 'mean' if col in df.select_dtypes(include='number').columns else 'first'
    for col in df.columns
    if col != 'TARGET_FID' and col != 'geometry'
}
groupedDf = df.groupby('TARGET_FID', as_index=False).agg(agg_rules)
groupedDf.head(5)
```

```python
groupedDf.info(verbose=True, show_counts=True)
```

```python
print("Dimension of the data: ", groupedDf.shape)

no_of_rows = groupedDf.shape[0]
no_of_columns = groupedDf.shape[1]

print(f'No. of Rows: {no_of_rows}')
print(f'No. of Columns: {no_of_columns}')
```

```python
groupedDfcopy = groupedDf.copy()
columns = groupedDfcopy.select_dtypes(include='number').columns
groupedDfcopy[columns] = groupedDfcopy[columns].mask(groupedDfcopy[columns] < -900)
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np

themePercentVars = groupedDfcopy.columns[groupedDfcopy.columns.str.startswith('EP_')][:16]
themePercent_TapVars = themePercentVars.tolist() + ['AvgDaily_T', 'AvgDaily_N', 'AvgDaily_E', 'NoTap_perc']

def plot_theme_histograms(data, variables):
    # Background groups follow the order of variables: 5, 5, 1, then the rest.
    background_colors = ['#e2f0d9'] * 5 + ['#fce4d6'] * 5 + ['#e4d7f5'] + ['#dceaf7'] * 5
    fig, axes = plt.subplots(
        nrows=(len(variables) + 3) // 4, ncols=4,
        figsize=(20, 15), squeeze=False
    )
    for i, (ax, column) in enumerate(zip(axes.flat, variables)):
        data[column].dropna().hist(bins=50, ax=ax)
        ax.set_title(column)
        ax.set_facecolor(background_colors[i] if i < 16 else '#382D2B')  # Tap Variables color
    for ax in list(axes.flat)[len(variables):]:
        ax.set_visible(False)
    legend_items = [
        Patch(facecolor='#e2f0d9', label='Socioeconomic Status'),
        Patch(facecolor='#fce4d6', label='Household Characteristics'),
        Patch(facecolor='#e4d7f5', label='Racial & Ethnic Minority Status'),
        Patch(facecolor='#dceaf7', label='Housing type & Transportation'),
        Patch(facecolor='#382D2B', label='Tap Variables')
    ]
    fig.legend(handles=legend_items, title='Background groups (themePercentVars order)',
               loc='upper center', ncol=4, frameon=True)
    fig.tight_layout(rect=[0, 0, 1, 0.94])
    plt.show()

plot_theme_histograms(groupedDfcopy, themePercent_TapVars)
```

```python
import numpy as np
from pandas.plotting import scatter_matrix


attributes = ["NoTap_perc", "EP_MINRTY", "EP_SNGPNT", "EP_HBURD", "EP_UNEMP", "EP_DISABL","EP_POV150"]

axes = scatter_matrix(groupedDfcopy[attributes], figsize=(15, 12))
correlations = groupedDfcopy[attributes].corr()

for row, y_column in enumerate(attributes):
    for col, x_column in enumerate(attributes):
        if row == col:
            continue  # Keep the diagonal histograms clear.
        r = correlations.loc[y_column, x_column]
        label = f'r = {r:.2f}' if np.isfinite(r) else 'r = N/A'
        axes[row, col].text(
            0.05, 0.95, label,
            transform=axes[row, col].transAxes,
            ha='left', va='top', fontsize=9,
            bbox=dict(facecolor='white', alpha=0.85, edgecolor='none')
        )

#Thie diagnol is either Histograms or Kernel Density Estimations <-- These are just plain old histograms

plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

model1 = LinearRegression()
model2 = LinearRegression()

# Use the same complete rows so the two models can be compared.
model_data = groupedDfcopy[['EP_MINRTY', 'EP_POV150', 'NoTap_perc']].dropna()
x1 = model_data[['EP_MINRTY']]
x2 = model_data[['EP_MINRTY', 'EP_POV150']]
y = model_data['NoTap_perc']

model1.fit(x1, y)
model2.fit(x2, y)

intercept1 = model1.intercept_
intercept2 = model2.intercept_
coefficients1 = model1.coef_
coefficients2 = model2.coef_
r_squared1 = model1.score(x1, y)
r_squared2 = model2.score(x2, y)

for name, model, features in [('Model 1', model1, x1), ('Model 2', model2, x2)]:
    print(f'\n{name} ({len(y)} rows)')
    print(f'Intercept: {model.intercept_:.4f}')
    print('Coefficients:', dict(zip(features.columns, model.coef_)))
    formula = f'NoTap_perc(predicted) = {model.intercept_:.4f}'
    for feature, coefficient in zip(features.columns, model.coef_):
        formula += f' {coefficient:+.4f} * {feature}'
    print('Prediction Formula:', formula)
    print(f'Training R?: {model.score(features, y):.4f}')
    print(f'Training RMSE: {root_mean_squared_error(y, model.predict(features)):.4f}')
```
